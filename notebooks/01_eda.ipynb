```python
caminho = '../'
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv(caminho+'database/raw/CBCovid19EC.csv')
df.head()
```

```python
# Distribuição de Positivos e negativos
df['pcr'].value_counts().plot(
    kind='bar',
    color=["#4C72B0", "#C44E52"],
    edgecolor="black",
    rot=0,
)

plt.title('Distribuição de Resultados')
plt.xlabel('Resultado')
plt.ylabel('Quantidade')
plt.show()
```

```python
df.isnull().sum()
```

```python
df_features = df.drop(['ID', 'pcr_date', 'exam_bh_date'], axis=1)
df_features.columns
```

```python
df['sex'].value_counts()
```

```python
sex_bin = [0 if sex == 'female' else 1 for sex in df['sex'].copy()]
df_features['sex_bin'] = sex_bin
```

```python
df_features['sex_bin'].value_counts()
```

```python
df_features.hist(figsize=(17, 13))
```

```python
df_features.plot(kind='box', subplots=True, figsize=(25, 10), sharey=False)
```

```python
# Eliminar Outliers
df_limpo = df_features[
    (df_features['monocytesP'] < 25) &
    (df_features['eosinophilsP'] < 15) &
    (df_features['neutrophils'] < 10000) &
    (df_features['lymphocytes'] < 6000) &
    (df_features['eosinophils'] < 1000) &
    (df_features['basophils'] < 100) &
    (df_features['mcv'] > 20) &
    (df_features['mch'] < 50) &
    (df_features['mchc'] < 80)
].copy()
```

```python
df_limpo.plot(kind='box', subplots=True, figsize=(25, 10), sharey=False)
```

```python
df_limpo.hist(figsize=(17, 13))
```

```python
print(f'Dataframe antes: {df.shape}')
print(f'Dataframe depois: {df_limpo.shape}')
```

```python
df_limpo = df_limpo.drop('sex', axis=1)
df_limpo.columns
```

```python
# Salvar conjunto de dados tratado
df_limpo.to_csv(caminho + 'database/processed/CBCovid19_Tratado.csv', index=False)
```

# Vizualização dos dados

```python
X = df_limpo.drop('pcr', axis=1).copy()
y = [0 if tipo == 'negative' else 1 for tipo in df_limpo['pcr']]
```

```python
from sklearn.preprocessing import StandardScaler
# Normalizar os dados

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
from sklearn.manifold import TSNE
import plotly.express as px
import pandas as pd

tsne = TSNE(n_components=3, perplexity=30, random_state=42)
X_tsne_3d = tsne.fit_transform(X_scaled)

df_tsne = pd.DataFrame(X_tsne_3d, columns=['TSNE1', 'TSNE2', 'TSNE3'])
df_tsne['classe'] = y  # ajuste se y não for array simples

fig = px.scatter_3d(
    df_tsne, x='TSNE1', y='TSNE2', z='TSNE3',
    color='classe',
    opacity=0.7,
    title='Visualização t-SNE 3D'
)
fig.update_traces(marker=dict(size=4))
fig.show()
```
