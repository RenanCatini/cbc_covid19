```python
caminho = '../'
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv(caminho+'database/raw/CBCovid19EC.csv')
df.head()
```

```python
df.describe()
```

```python
df.dtypes
```

```python
# Distribuição de Positivos e negativos
df['pcr'].value_counts().plot(
    kind='bar',
    color=["#4C72B0", "#C44E52"],
    edgecolor="black",
    rot=0,
)

plt.title('Distribuição de Resultados')
plt.xlabel('Resultado')
plt.ylabel('Quantidade')
plt.show()
```

```python
df.isnull().sum()
```

```python
df_features = df.drop(['ID', 'pcr_date', 'exam_bh_date', 'sex', 'age'], axis=1)
df_features.columns
```

```python
df_features.hist(figsize=(17, 13))
```

```python
df_features.plot(kind='box', subplots=True, figsize=(25, 10), sharey=False)
```

```python
# Eliminar Outliers
df_limpo = df_features[
    (df_features['monocytesP'] < 25) &
    (df_features['eosinophilsP'] < 15) &
    (df_features['neutrophils'] < 10000) &
    (df_features['lymphocytes'] < 6000) &
    (df_features['eosinophils'] < 1000) &
    (df_features['basophils'] < 100) &
    (df_features['mcv'] > 20) &
    (df_features['mch'] < 50) &
    (df['mchc'] < 80)
].copy()
```

```python
df_limpo.plot(kind='box', subplots=True, figsize=(25, 10), sharey=False)
```

```python
df_limpo.hist(figsize=(17, 13))
```

```python
print(f'Dataframe antes: {df.shape}')
print(f'Dataframe depois: {df_limpo.shape}')
```

```python
# Salvar conjunto de dados tratado
df_limpo.to_csv(caminho + 'database/processed/CBCovid19_Tratado.csv', index=False)
```
